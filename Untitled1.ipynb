```python
# Instalar una sola vez, si hace falta:
# pip install pandas numpy scikit-learn matplotlib

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import confusion_matrix, accuracy_score, roc_curve, roc_auc_score

# 1. Cargar y preparar datos
votos = pd.read_csv("house-votes-84.data", header=None, na_values="?")
votos.columns = ["clase"] + [f"V{i}" for i in range(1, 17)]
print(votos.shape)
print(votos["clase"].value_counts())

# Codificar: n = 0, y = 1, NA = 2 (categoría propia, ver nota abajo)
# Clase: republican = 0, democrat = 1 (positivo = democrat)
X = votos.drop(columns="clase").replace({"n": 0, "y": 1}).fillna(2).astype(int)
y = (votos["clase"] == "democrat").astype(int)

# 2. Separar entrenamiento (80 %) y prueba (20 %)
X_ent, X_prue, y_ent, y_prue = train_test_split(
    X, y, train_size=0.8, random_state=42
)

# 3. Entrenar (la clase NO está entre los predictores)
modelo = CategoricalNB(alpha=1, min_categories=3)
modelo.fit(X_ent, y_ent)

# 4. Predicciones y matriz de confusión
clase_pred = modelo.predict(X_prue)
matriz = pd.DataFrame(
    confusion_matrix(y_prue, clase_pred, labels=[0, 1]),
    index=pd.Index(["republican", "democrat"], name="Real"),
    columns=pd.Index(["republican", "democrat"], name="Predicha"),
)
print(matriz)
exactitud = accuracy_score(y_prue, clase_pred)
print("Exactitud:", round(exactitud, 4))

# 5. Probabilidad de la clase positiva (democrat)
prob = modelo.predict_proba(X_prue)[:, 1]

# 6. ROC y AUC: positivo = democrat
fpr, tpr, _ = roc_curve(y_prue, prob)
auc = roc_auc_score(y_prue, prob)
plt.plot(fpr, tpr, color="blue", lw=2, label=f"AUC = {auc:.4f}")
plt.plot([0, 1], [0, 1], "--", color="gray")
plt.xlabel("1 - Especificidad (FPR)")
plt.ylabel("Sensibilidad (TPR)")
plt.title("Curva ROC - Naive Bayes")
plt.legend()
plt.show()
print("AUC:", auc)

# 7. Analizar errores
errores = votos.loc[X_prue.index[clase_pred != y_prue]]
print(errores)
```
